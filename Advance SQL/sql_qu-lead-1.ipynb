```python
%sql
select * from (
select 
emp_name, salary, row_number() over (partition by emp_name order by salary_date desc) as rn, lag(salary) over (partition by emp_name order by salary_date) prev_sal from test.genie_demo.emp_salary ) where rn=1
```

```python
%sql
CREATE TABLE test.genie_demo.emp_salary_hist (
    emp_id INT,
    salary INT,
    effective_date DATE
);
```

```python
%sql
INSERT INTO test.genie_demo.emp_salary_hist VALUES
(1, 50000, '2024-01-01'),
(1, 50000, '2024-02-01'),
(1, 55000, '2024-03-01'),
(1, 55000, '2024-04-01'),
(2, 40000, '2024-01-01'),
(2, 42000, '2024-02-01');
```

```python
%sql
select * from test.genie_demo.emp_salary_hist
```

```python
%sql
select emp_id, salary,effective_date from (
select emp_id, salary,effective_date, case when
(salary - lag(salary) over (partition by emp_id order by effective_date)) != 0 then 'Y' else 'N' end as flag
 from test.genie_demo.emp_salary_hist ) where flag='Y'  
```

```python
%sql
select emp_id, salary, effective_date
from (
  select *,
         lag(salary) over (partition by emp_id order by effective_date) as prev_salary
  from test.genie_demo.emp_salary_hist
)
where prev_salary is null OR salary <> prev_salary
```
