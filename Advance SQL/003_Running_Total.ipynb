### First Day When Customer Crossed Spending Threshold - Return the first transaction date when a customer's cumulative spend became greater than or equal to 500.

```python
%sql
CREATE OR REPLACE TEMP VIEW customer_transactions AS
SELECT *
FROM VALUES
(1, '2025-01-01', 100),
(1, '2025-01-03', 200),
(1, '2025-01-05', 400),
(1, '2025-01-07', 300),

(2, '2025-01-01', 150),
(2, '2025-01-02', 100),
(2, '2025-01-04', 200),

(3, '2025-01-01', 500),
(3, '2025-01-03', 600)

AS customer_transactions(customer_id, txn_date, amount);
```

```python
%sql
select * from customer_transactions
```

```python
%sql
WITH running_total AS (
    SELECT
        customer_id,
        txn_date,
        SUM(amount) OVER (
            PARTITION BY customer_id
            ORDER BY txn_date
            ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
        ) AS running_total
    FROM customer_transactions
)
SELECT
    customer_id,
    MIN(txn_date) AS first_cross_date
FROM running_total
WHERE running_total >= 500
GROUP BY customer_id;
```
