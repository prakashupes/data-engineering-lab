```python
from pyspark.sql import Row

# a tiny sample DataFrame
df = spark.createDataFrame([
    Row(post_id="post-1", type="view", user_id="u10", ts=1790871216),
    Row(post_id="post-1", type="like", user_id="u11", ts=1790871217),
    Row(post_id="post-2", type="view", user_id="u12", ts=1790871218),
])

base = "s3://retail-data-lake-dev/kafka/poc_formats"

# write as Parquet (columnar)
df.write.mode("overwrite").format("parquet").save(f"{base}/parquet")

# write as Avro (row-based)
df.write.mode("overwrite").format("avro").save(f"{base}/avro")

print("written to:", base)
```
