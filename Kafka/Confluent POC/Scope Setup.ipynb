```python
import requests

# These two are auto-available inside any Databricks notebook — no token needed
ctx   = dbutils.notebook.entry_point.getDbutils().notebook().getContext()
host  = ctx.apiUrl().get()
token = ctx.apiToken().get()

headers = {"Authorization": f"Bearer {token}"}

# 1. Create the scope
r = requests.post(f"{host}/api/2.0/secrets/scopes/create",
                  headers=headers,
                  json={"scope": "kafka_lab"})
print("create-scope:", r.status_code, r.text)
```
