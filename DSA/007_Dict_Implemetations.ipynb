## Create a set, which have insert, remove in O(1)

```python
import random
class RandomizedSet:

    def __init__(self):
        self.lst_index= {}
        self.lst = []

        

    def insert(self, val: int) -> bool:
        res = False
        index = self.lst_index.get(val) 
        length = len(self.lst)
        if (index is None) or (index==-1):
            self.lst.append(val)
            self.lst_index[val]=length
            res= True
        return res

        

    def remove(self, val: int) -> bool:
        res = False
        index = self.lst_index.get(val) 
        length = len(self.lst)
        if (index is not None):
            if(index!=-1):
                self.lst_index[self.lst[length-1]]=index
                self.lst[index], self.lst[length-1] = self.lst[length-1], self.lst[index]
                self.lst.pop()
                self.lst_index[val]=-1
                res= True
        return res
        
        

    def getRandom(self) -> int:
        return random.choice(self.lst)
        


```

```python
obj = RandomizedSet()
print(obj.insert(1))
print(obj.lst, obj.lst_index)
print(obj.insert(5))
print(obj.lst, obj.lst_index)



# param_2 = obj.remove(val)
# param_3 = obj.getRandom()
```
