### A matrix uses **`matrix[row][col]`**. 
- Moving up i-- , moving down i++ --> changes the row;
- moving left j--, moving right j++ --> changes the column.

```python
metrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print(metrix)
```

```python
### Qu: Given location is (2,1), print its all above, below, left , right
```

```python
i,j = 2,1
print('location ', metrix[i][j])
print('left ', metrix[i][j-1])
print('right ', metrix[i][j+1])
print('Above ', metrix[i-1][j])
##print('Below ', metrix[i+1][j])
```

```python
## Print Right Diagonal Op: 3,5,7

i=0
j=len(metrix[0])-1

while i<len(metrix) and j>=0:
    print(metrix[i][j])
    i+=1
    j-=1
## Print Left Diagonal Op: 1,)
##
```
