### Indexing: 
a = [1, 2, 3, 4]
-     0  1  2  3     <- normal index
-    -4 -3 -2 -1     <- negative index

**arr[start:end:step] : end is exclusive** 

Default Values-  arr[:::] when +ve direction is given 
start = 0 ,end = len(arr) ,step = 1 

Arr[::-1], if negative direction is given: 

start = -1 ,end = -(len(arr)+1) or -len(arr)-1 
step = -1 

```python
a= [5,3,4,6]
b = a[::-1]
a.reverse()
print(a)

print(b)
```

### Comprehansion
- A compact way to create lists in one line. 
- Syntax: [expression for item in iterable] 

```python
lst = [i*2 for i in range(10)]
print(lst)
```

```python
arr = [2,3,4,5,6]
even = [x for x in arr if x%2==0]
print(even)
```

```python
arth = lambda x,y,z : x*y+z

print(arth(3,4,5))

### lst comprehension using lambda

lst = [3,4,5,6]
res = [lambda(x: x%2==0) (X) for x in lst]
print(res)
```
