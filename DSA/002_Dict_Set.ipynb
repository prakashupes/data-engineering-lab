```python
d = {"apple": 100, "banana": 200}

# Look up a value
print(d["apple"])                 # 100

# Insert a new key-value pair
d["mango"] = 300

# Update an existing key
d["banana"] = 250

# Check whether a key exists
print("banana" in d)              # True
print("orange" in d)              # False

# Safe lookup: return a default if the key is missing
print(d.get("orange", 0))         # 0

```
