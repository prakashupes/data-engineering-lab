## Create a set, which have insert, remove in O(1) - Set: At a time only 1 occurance should be present

```python
import random
class RandomizedSet:

    def __init__(self):
        self.lst_index= {}
        self.lst = []

        

    def insert(self, val: int) -> bool:
        res = False
        index = self.lst_index.get(val) 
        length = len(self.lst)
        if (index is None) or (index==-1):
            self.lst.append(val)
            self.lst_index[val]=length
            res= True
        return res

        

    def remove(self, val: int) -> bool:
        res = False
        index = self.lst_index.get(val) 
        length = len(self.lst)
        if (index is not None):
            if(index!=-1):
                self.lst_index[self.lst[length-1]]=index
                self.lst[index], self.lst[length-1] = self.lst[length-1], self.lst[index]
                self.lst.pop()
                self.lst_index[val]=-1
                res= True
        return res
        
        

    def getRandom(self) -> int:
        return random.choice(self.lst)
        


```

### Insert Delete GetRandom O(1) - Duplicates allowed

- RandomizedCollection() Initializes the empty RandomizedCollection object.
- bool insert(int val) Inserts an item val into the multiset, even if the item is already present. Returns true if the item is not present, false otherwise.
- bool remove(int val) Removes an item val from the multiset if present. Returns true if the item is present, false otherwise. Note that if val has multiple occurrences in the multiset, we only remove one of them.
- int getRandom() Returns a random element from the current multiset of elements. The probability of each element being returned is linearly related to the number of the same values the multiset contains.
- You must implement the functions of the class such that each function works on average O(1) time complexity.

```python

import random

class RandomizedCollection:

    def __init__(self):
        self.mutliset = []
        self.mutliset_index = {}

    def insert(self, val: int) -> bool:
        is_new = val not in self.mutliset_index

        if is_new:
            self.mutliset_index[val] = set()

        self.mutliset_index[val].add(len(self.mutliset))
        self.mutliset.append(val)

        return is_new

    def remove(self, val: int) -> bool:
        if val not in self.mutliset_index:
            return False

        # Get and remove one occurrence's index
        val_index = self.mutliset_index[val].pop()

        # Identify the last element
        last_index = len(self.mutliset) - 1
        last_val = self.mutliset[last_index]

        # Move the last element into the removed position
        if val_index != last_index:
            self.mutliset[val_index] = last_val

            # Update the moved element's index set
            self.mutliset_index[last_val].remove(last_index)
            self.mutliset_index[last_val].add(val_index)

        # Remove the last element
        self.mutliset.pop()

        # Delete the key if no occurrences remain
        if not self.mutliset_index[val]:
            del self.mutliset_index[val]

        return True

    def getRandom(self) -> int:
        return random.choice(self.mutliset)
```
